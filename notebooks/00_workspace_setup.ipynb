```python
from pathlib import Path

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/image-captioning-rag"
)

RAW_ROOT = DRIVE_ROOT / "00_original_data"
MANIFEST_ROOT = DRIVE_ROOT / "01_manifests"
PROCESSED_ROOT = DRIVE_ROOT / "02_processed"
EMBEDDINGS_ROOT = DRIVE_ROOT / "03_embeddings"
INDEX_ROOT = DRIVE_ROOT / "04_indexes"
REPORTS_ROOT = DRIVE_ROOT / "05_reports"
RELEASE_ROOT = DRIVE_ROOT / "06_releases"

for path in [
    MANIFEST_ROOT,
    PROCESSED_ROOT,
    EMBEDDINGS_ROOT,
    INDEX_ROOT,
    REPORTS_ROOT,
    RELEASE_ROOT,
]:
    path.mkdir(parents=True, exist_ok=True)

print("Drive connected:", DRIVE_ROOT.exists())
print("Project folder:", DRIVE_ROOT)
```
