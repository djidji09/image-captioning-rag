# Phase One: audit, candidate selection, and frozen splits

This notebook delegates all work to `scripts/phase1.py`; it does not duplicate pipeline code. Configure dataset locations in `configs/phase1.yaml` first. The default paths assume the COCO directories are siblings of this repository.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
SCRIPT = ROOT / "scripts" / "phase1.py"
CONFIG = ROOT / "configs" / "phase1.yaml"

def run_phase1(command, *args):
    subprocess.run([sys.executable, str(SCRIPT), command, "--config", str(CONFIG), *args], cwd=ROOT, check=True)
```

Run the source-data audit before selecting candidates. The selection step hashes and validates the sampled image files and may take several minutes.

```python
run_phase1("audit")
run_phase1("select")
```

Freeze the split only after candidate selection succeeds. Re-running refuses to overwrite an existing frozen split; use `--force-new-splits` deliberately to archive and replace it.

```python
run_phase1("split")
split_manifest = ROOT / "01_splits" / "splits_v2.json"
print(split_manifest)
```
